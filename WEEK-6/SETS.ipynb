```python
s1={1,2,3}
s2={1,2,3}
print(s1|s2)
print(s1&s2)
print(s1-s2)
print(s1^s2)
```

```python
#To change the original set
c={11,22,33}
a={1,2,3,33}
print(a.union(c))
print(a.difference(c))
print(a.intersection(c))
print(a.intersection_update(c))
print(a) # a is intersection set contains only 33
```
