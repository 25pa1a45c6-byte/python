```python

t1=(4,5,2,7)
t2=(3,7,2,9)
print(t1+t2)#concatenation
print(t1*2)#repetition
print(t2*3)#repetition
print(5 in t1)#membership
print(7 not in t1)
print(t1[2])#indexing
print(t1[::-1])#slicing
print(t1[1:3])

```

```python
print(sum(t1))
print(sum(t2))
```

```python
t1=(3,1,6,4,8,9)
t2=(4,5,23,2,65,7)
print(max(t1))
print(min(t1))
print(len(t1))
print(sorted(t1))
print(tuple(zip(t1,t2)))
```
