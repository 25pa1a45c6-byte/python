```python
#1.finding greatest among three numbers
a=int(input("enter a number for a "))
b=int(input("enter a number for b "))
c=int(input("enter a number for c "))
if a>b and a>c:
   print(f"{a} is greater")
elif a==b and a==c:
   print("three  are equal")
elif b>a and b>c:
   print(f"{b} is greater")
else:
   print(f"{c} is greater")
```

```python
#2.print fizz if a  number is divisible by 3,print buzz if an number is divisible by 5 ,print fizzbuzz if it is divisible by both 3 and 5
num=int(input("enter a number "))
if num%3==0 and num%5!=0:
  print("FIZZ")
elif num%3!=0 and num%5==0:
  print("BUZZ")
elif num%3==0 and num%5==0:
  print("FIZZBUZ")
else:
  print ("nothing")

```

```python
#3.addition and multipilcation of two complex numbers
a=complex(input("enter 1st complex num:"))
b=complex(input("enter 2nd complex num:"))
print(a+b)
print(a*b)
```

```python
#4.program to swap 2 variables without using temp variable
a=int(input("enter first number:"))
b=int(input("enter second number:"))
print(f"{a,b} before swapping")
a,b=b,a
print(f"{a,b} after swapping")
```

```python
#5.program on match
day=int(input("enter a number:"))
match day:
  case 1:
    print("sunday")
  case 2:
    print("monday")
  case 3:
    print("tuesday")
  case 4:
    print("wednesday")
  case 5:
    print("thursday")
  case 6:
    print("friday")
  case 7:
    print("saturday")
  case _:
    print("invalid choice")
```
