```python
#program that checks whether given number is palindrome or not using functions
def palindrome(num):
    original = num
    reverse = 0

    while num > 0:
        digit = num % 10
        reverse = reverse * 10 + digit
        num //= 10

    if original == reverse:
      print("palindrome")
    else:
      print("not a palindrome")


n = int(input("Enter a number: "))
palindrome(n)
```

```python
#program that reverse a given string using function
def reverse(s):
  rev=""
  for i in range(len(s)-1,-1,-1):
    rev=rev+s[i]
  print(rev)
s=input()
reverse(s)
```

```python
#another method
def reverse(s):
  return s[::-1]
s=input()
print(reverse(s))
```

```python
#functio0n that returns multiple values
def multiple_values(*num):
  s=sum(num)
  c=len(num)
  avg=s/c
  return s,c,avg
res_sum,res_count,res_avg=multiple_values(1,2,3,4,5)
print(f"the sum is ={res_sum},the count is= {res_count},the average is= {res_avg}")
```
