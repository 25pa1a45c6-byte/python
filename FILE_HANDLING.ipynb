```python
file=open("data.txt","w")
contest="1 new line"
file.write(contest)
file.close()
```

```python
file=open("data.txt","w")
file.write("new line 1")
file.close()
```

```python
with open("data.txt","a") as file:
  file.write("new line 2")
```

```python
with open("data.txt","a") as file:
  file.writelines(["\n new line 3","\n new line 4"])
```

```python
with open("data.txt","r") as file:
  print(file.read())
```

```python
with open("data.txt","r") as file:
  content=file.read(10)
  print(content)
  print(file.tell())
  file.seek(0)
  print(file.read())
```

```python
with open("data.txt","r") as file:
  print(file.readline())
  print(file.readline())
```

```python
with open("data.txt","r") as file:
  print(file.readlines())
```
