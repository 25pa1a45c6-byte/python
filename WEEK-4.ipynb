```python
#program that count no of vowels and consonants in a string
```

```python
a=str(input())
b=a.lower()
vowels="aeiou"
i=0
j=0
for ch in b:
  if ch.isalpha():
    if ch in vowels:
      i+=1
    else:
      j+=1
print("the count of vowels is",i)
print("the count 0f consonants is",j)

```
