```python
#Class and Object
class Student:
    name = "akash"
    age = 20
s1 = Student()
print(s1.name)
print(s1.age)
```

```python
#Constructor
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age
s1 = Student("akash", 20)
print(s1.name)
print(s1.age)
```

```python
class student:
  def __init__(self, name ,marks):
     self.name= name
     self.marks= marks
  def display(self):
    print(self.name,self.marks)
s1=student("AKASH",45)
s2=student("ravi",67)
print(s1.name,s1.marks)
print(s2.name,s2.marks)
```

```python
class Calculator:
    def add(self, a, b):
        return a + b
    def sub(self, a, b):
        return a - b
c = Calculator()
print(c.add(10, 5))
print(c.sub(10, 5))
```
